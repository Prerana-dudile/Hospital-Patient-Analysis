```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"D:\Datasets\hospital_patient_data_cleaned.csv")
```

```python
df.shape
```

# Q1.Which medical conditions are most common among patients?

```python
condition_counts = df['Condition'].value_counts()

plt.figure(figsize=(10, 6))
condition_counts.plot(kind='bar')
plt.title('Number of Patients by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Number of Patients')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The chart shows the distribution of patients across different medical conditions. The most common condition was Fractured Leg and Heart Attack, while Osteoarthritis had the lowest number of patients.

# Q2.What is the gender distribution of patients?

```python
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(7, 5))
gender_counts.plot(kind='bar')
plt.title('Patient Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Patients')
plt.tight_layout()
plt.show()
```

# Q3.Which states have the highest number of patients?

```python
state_counts = df['Patient_State'].value_counts()

plt.figure(figsize=(12, 6))
state_counts.plot(kind='bar')
plt.title('Number of Patients by State')
plt.xlabel('State')
plt.ylabel('Number of Patients')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

# Q4.What is the age distribution of hospital patients?

```python
plt.figure(figsize=(10, 6))
plt.hist(df['Age'], bins=10)
plt.title('Age Distribution of Patients')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.tight_layout()
plt.show()
```

# Q5.How long do patients typically stay in the hospital?

```python
plt.figure(figsize=(10, 6))

plt.hist(df['Length_of_Stay'], bins=10)

plt.title('Distribution of Length of Hospital Stay')
plt.xlabel('Length of Stay (Days)')
plt.ylabel('Number of Patients')

plt.tight_layout()
plt.show()
```

# Q6.Which medical conditions require longer hospital stays?

```python
avg_stay = df.groupby('Condition')['Length_of_Stay'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
avg_stay.plot(kind='bar')
plt.title('Average Length of Stay by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Average Length of Stay (Days)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Q7.Which medical conditions have the highest total treatment costs?

```python
total_cost = df.groupby('Condition')['Total_Cost'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
total_cost.plot(kind='bar')
plt.title('Total Treatment Cost by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Total Treatment Cost')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Q8.Which medical conditions have the highest average treatment cost per patient?

```python
avg_cost = df.groupby('Condition')['Total_Cost'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
avg_cost.plot(kind='bar')
plt.title('Average Treatment Cost by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Average Treatment Cost')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Q11.How many patients claimed insurance?

# Q12.Does a longer hospital stay tend to be associated with higher treatment costs?

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Length_of_Stay'], df['Total_Cost'])
plt.title('Length of Stay vs Treatment Cost')
plt.xlabel('Length of Stay (Days)')
plt.ylabel('Total Treatment Cost')
plt.tight_layout()
plt.show()
```

# Q13.How satisfied are patients with their hospital experience?

```python
satisfaction_counts = df['Satisfaction'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
satisfaction_counts.plot(kind='bar')
plt.title('Patient Satisfaction Distribution')
plt.xlabel('Satisfaction Score')
plt.ylabel('Number of Patients')
plt.tight_layout()
plt.show()
```

# Q14.Which medical conditions have the highest average patient satisfaction?

```python
avg_satisfaction = df.groupby('Condition')['Satisfaction'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
avg_satisfaction.plot(kind='bar')
plt.title('Average Patient Satisfaction by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Average Satisfaction')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Q15.Age Group Analysis

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 39, 59, 100],
    labels=['Young', 'Middle Age', 'Senior']
)

age_group_counts = df['Age_Group'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
age_group_counts.plot(kind='bar')
plt.title('Patients by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Patients')
plt.tight_layout()
plt.show()
```
